```python

import random

tag_prob = {
    "book": {"NN": 0.8, "VB": 0.2},
    "run": {"NN": 0.3, "VB": 0.7},
    "the": {"DT": 1.0}
}

words = ["the", "book", "run"]

for word in words:
    tags = list(tag_prob[word].keys())
    probabilities = list(tag_prob[word].values())
    tag = random.choices(tags, weights=probabilities)[0]
    print(word, ":", tag)
```
