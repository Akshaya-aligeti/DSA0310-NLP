```python

import re

def tag_word(word):
    if re.match(r".*ing$", word):
        return "Verb"
    elif re.match(r".*ly$", word):
        return "Adverb"
    elif re.match(r".*ous$", word):
        return "Adjective"
    else:
        return "Noun"

words = ["playing", "quickly", "beautiful", "apple"]

for word in words:
    print(word, ":", tag_word(word))
```
