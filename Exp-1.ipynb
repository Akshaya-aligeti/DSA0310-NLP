```python

import re

text = "Python is easy to learn. Contact us at 9876543210"

# Match a pattern at the beginning
result1 = re.match("Python", text)
print("Match:", result1.group() if result1 else "Not Found")

# Search for a pattern anywhere in the text
result2 = re.search("easy", text)
print("Search:", result2.group() if result2 else "Not Found")

# Find all numbers in the text
result3 = re.findall(r"\d+", text)
print("Find All:", result3)

# Check whether a pattern exists
result4 = re.search(r"\d{10}", text)
print("Mobile Number:", result4.group() if result4 else "Not Found")
```
