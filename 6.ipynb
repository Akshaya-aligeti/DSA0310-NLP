```python

import random

text = "I love Python. I love programming."
words = text.replace(".", "").split()

bigrams = {}

for i in range(len(words) - 1):
    word = words[i]
    next_word = words[i + 1]
    bigrams.setdefault(word, []).append(next_word)

word = "I"
result = [word]

for i in range(5):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print("Generated Text:", " ".join(result))
```
