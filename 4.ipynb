```python

def plural(noun):
    if noun.endswith(("s", "sh", "ch", "x")):
        return noun + "es"
    elif noun.endswith("y"):
        return noun[:-1] + "ies"
    else:
        return noun + "s"

word = input("Enter a noun: ")
print("Plural form:", plural(word))
```
