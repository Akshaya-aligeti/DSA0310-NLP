```python

import re

expression = "Human(x) AND Mortal(x)"

pattern = r"([A-Za-z]+)\((\w+)\)"

matches = re.findall(pattern, expression)

print("Logical Expression:", expression)

for predicate, variable in matches:
    print("Predicate:", predicate)
    print("Variable:", variable)
```
