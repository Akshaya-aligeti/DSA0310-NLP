```python

import re

text = "Ravi bought a car. He drives it daily."

nouns = ["Ravi", "car"]
pronouns = ["He", "it"]

for pronoun, noun in zip(pronouns, nouns):
    if re.search(r"\b" + pronoun + r"\b", text):
        print(pronoun, "refers to", noun)
```
