```python

words = ["They", "play", "cricket"]

tags = ["NN", "NN", "NN"]

for i in range(len(words)):
    if words[i] == "They":
        tags[i] = "PRP"
    elif words[i] == "play":
        tags[i] = "VBP"
    elif words[i] == "cricket":
        tags[i] = "NN"

for word, tag in zip(words, tags):
    print(word, ":", tag)
```
