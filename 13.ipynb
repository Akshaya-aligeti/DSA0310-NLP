```python

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> 'Ram'
VP -> V NP
V -> 'eats'
NP -> 'apple'
""")

parser = nltk.ChartParser(grammar)

sentence = "Ram eats apple".split()

for tree in parser.parse(sentence):
    tree.pretty_print()
```
